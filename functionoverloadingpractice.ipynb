```python
def sub(a,b):
    return a-b
print(sub(50,10))
```

```python
def add(a,b,c=0):
    return a+b+c
print(add(10,30))
print(add(10,30,50))
```

```python
def mul(a,b):
    return a*b
print(mul(5,9))
```

```python
def mul(a,b,c=0):
    return a*b*c
print(mul(2,3))
print(mul(14,15,16))
```

```python
def add(*numbers):
    print(sum(numbers))
add(10,20)
add(10,20,30)
add(10,20,30,40)
```

```python
def student(*marks):
    print(sum(marks))
student(72,98)
student(78,88,99)
student(69,68,67,66)
```
